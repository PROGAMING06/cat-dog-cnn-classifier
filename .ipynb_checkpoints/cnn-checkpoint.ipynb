```python
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader, random_split
```

```python
# training transform: includes augmentation
train_transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.RandomHorizontalFlip(),          # 50% chance to flip image left-right
    transforms.RandomRotation(10),               # randomly rotate up to 10 degrees
    transforms.ColorJitter(brightness=0.2),      # randomly adjust brightness slightly
    transforms.ToTensor(),
    transforms.Normalize((0.5,0.5,0.5), (0.5,0.5,0.5))
])

# test transform: NO augmentation — just resize/normalize, must match training's final format
test_transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,0.5,0.5), (0.5,0.5,0.5))
])
```

```python
# load twice — same images, different transforms
full_dataset_train = datasets.ImageFolder(root='C:/Users/HP/all_projects/CNN/training_set', transform=train_transform)
full_dataset_test = datasets.ImageFolder(root='C:/Users/HP/all_projects/CNN/training_set', transform=test_transform)

# generate split indices once, using a fixed seed for reproducibility
torch.manual_seed(42)
train_size = int(0.8 * len(full_dataset_train))
test_size = len(full_dataset_train) - train_size
train_indices, test_indices = random_split(range(len(full_dataset_train)), [train_size, test_size])

from torch.utils.data import Subset
train_data = Subset(full_dataset_train, train_indices.indices)
test_data = Subset(full_dataset_test, test_indices.indices)

train_loader = DataLoader(train_data, batch_size=32, shuffle=True)
test_loader = DataLoader(test_data, batch_size=32, shuffle=False)

print('train images:', len(train_data))
print('test images:', len(test_data))
```

```python
# implementing dropout
import torch.nn as nn

class CatDogCNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.conv1 = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, padding=1)
        self.relu1 = nn.ReLU()
        self.pool1 = nn.MaxPool2d(kernel_size=2)

        self.conv2 = nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3, padding=1)
        self.relu2 = nn.ReLU()
        self.pool2 = nn.MaxPool2d(kernel_size=2)

        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(32 * 16 * 16, 64)
        self.relu3 = nn.ReLU()
        self.dropout = nn.Dropout(p=0.5)     # new
        self.fc2 = nn.Linear(64, 2)

    def forward(self, x):
        x = self.pool1(self.relu1(self.conv1(x)))
        x = self.pool2(self.relu2(self.conv2(x)))
        x = self.flatten(x)
        x = self.relu3(self.fc1(x))
        x = self.dropout(x)                   # new
        x = self.fc2(x)
        return x

model = CatDogCNN()
print(model)
```

```python
import torch

criterion = nn.CrossEntropyLoss()   # standard loss for multi-class classification (cat vs dog = 2 classes)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
epochs = 15

for epoch in range(epochs):
    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in train_loader:
        # forward pass
        outputs = model(images)
        loss = criterion(outputs, labels)

        # backward pass
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        # track stats
        running_loss += loss.item()
        _, predicted = torch.max(outputs, 1)
        correct += (predicted == labels).sum().item()
        total += labels.size(0)

    epoch_loss = running_loss / len(train_loader)
    epoch_acc = 100 * correct / total
    print(f'Epoch {epoch+1}/{epochs}, Loss: {epoch_loss:.4f}, Accuracy: {epoch_acc:.2f}%')
```

```python
# to test with the unseen dataset
model.eval()

correct = 0
total = 0

with torch.no_grad():
    for images, labels in test_loader:
        outputs = model(images)
        _, predicted = torch.max(outputs, 1)
        correct += (predicted == labels).sum().item()
        total += labels.size(0)

test_acc = 100 * correct / total
print(f'Test Accuracy: {test_acc:.2f}%')
```
